```python
class_count = df["Class"].value_counts()

plt.pie(
    class_count,
    labels=class_count.index,
    autopct="1%.1f%%",
    startangle=90,
    explode = [0.05] * len(class_count),
    shadow=True,
    colors= plt.cm.Set2.colors
)
plt.title("Distribution of Classes",fontsize = 16,fontdict={"fontweight":"bold"})
plt.axis("equal")
plt.ylabel("")
plt.show()
```

```python
df1 = df.drop("Class",axis=1)
```

```python
plt.figure(figsize=(8,6))
corr = df1.corr()

sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5,
    linecolor="white",
    square=True,
    cbar_kws={"shrink":.8},
    annot_kws={"size":10}
)
plt.title("Correlation Heatmap",fontsize=16,fontdict={"fontweight":"bold"})
plt.xticks(rotation = 45,ha = "right")
plt.yticks(rotation = 0)
plt.tight_layout()
plt.show()
```
